## 案例 1.1 \
缺失情况下默认 requires_grad == False

```python

a = torch.randn(2, 2) 
a = ((a * 3) / (a - 1))
print(a.requires_grad)
```

## 案例 1.2 \
requires_grad=True

```python
# 初始创建 .grad_fn 属性为 None
x = torch.randn(3, 3, requires_grad=True)
print(x.grad_fn)
```

```python
y = x**2
print(y)
print(y.grad_fn)
```

```python
z = y * y * 3
out = z.mean()

print(z, out)
print(z.grad_fn, out.grad_fn)
```
