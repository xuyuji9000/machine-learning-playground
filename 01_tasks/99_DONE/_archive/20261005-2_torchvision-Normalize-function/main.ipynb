```python
import torch
from torchvision import transforms
from PIL import Image
```

```python
# 1. 定义预处理/数据增强管道 (Pipeline)
# 使用 transforms.Compose 将多个操作串联起来
transform_pipeline = transforms.Compose([
    # 步骤 1: 调整图像尺寸为 256x256
    transforms.Resize((256, 256)),
    
    # 步骤 2: 训练时的随机水平翻转（如果是测试/验证集可以去掉这行）
    transforms.RandomHorizontalFlip(p=0.5),
    
    # 步骤 3: 从中心裁剪出 224x224 的区域
    transforms.CenterCrop(224),
    
    # 步骤 4: 转换为 Tensor，并将像素值归一化到 [0, 1]
    transforms.ToTensor(),
    
    # 步骤 5: 标准化（使用经典 ImageNet 的均值和标准差）
    # 公式: output = (input - mean) / std
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406], 
        std=[0.229, 0.224, 0.225]
    )
])
```

```python

# 2. 模拟加载一张图片（这里用纯色图像代替实际文件读取）
# 实际开发中通常使用: image = Image.open("path_to_image.jpg").convert("RGB")

image = Image.new("RGB", (300, 300), color="blue")
```

```python

# 3. 对图像应用定义好的 transform 管道
tensor_image = transform_pipeline(image)

# 4. 查看转换后的结果信息
print(f"转换后的类型: {type(tensor_image)}")
print(f"张量形状 (Channels, Height, Width): {tensor_image.shape}")
print(f"张量数据类型: {tensor_image.dtype}")
print(f"张量数值范围: [min: {tensor_image.min():.4f}, max: {tensor_image.max():.4f}]")
```
