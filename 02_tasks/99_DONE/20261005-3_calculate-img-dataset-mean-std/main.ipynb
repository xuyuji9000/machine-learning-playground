```python
import torch
from torch.utils.data import DataLoader
from torchvision import datasets, transforms
```

```python
# 1. 定义基本的转换（只转成 Tensor，先不进行 Normalize）
transform = transforms.Compose([
    transforms.Resize((224, 224)),  # 统一尺寸
    transforms.ToTensor()           # 将像素缩放到 [0, 1]
])
```

```python
# 2. 加载 FashionMNIST 数据集
dataset = datasets.FashionMNIST(root='./data', train=True, download=True, transform=transform)
dataloader = DataLoader(dataset, batch_size=64, shuffle=False, num_workers=4)

print(f"Dataset size: {len(dataset)}")

```

```python
# 3. Calculate global mean and std
train_mean = dataset.data.float().mean() / 255.0
train_std = dataset.data.float().std() / 255.0

print(f'Mean: {train_mean:.4f}')
print(f'Std: {train_std:.4f}')
```
