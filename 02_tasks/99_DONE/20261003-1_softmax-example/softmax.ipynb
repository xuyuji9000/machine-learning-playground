```python
import math
import numpy as np
```

```python
def softmax_py(xs):
    """Pure-Python softmax with max subtraction for numerical stability."""
    m = max(xs)
    exps = [math.exp(x - m) for x in xs]
    s = sum(exps)
    return [e / s for e in exps]
```

```python
# calculate softmax
logits = [1.0, 2.0, 3.0, 4.0, 1000.0]  # 1000 tests stability
print("pure python:", softmax_py(logits))
```
