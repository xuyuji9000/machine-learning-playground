```python
from torchviz import make_dot
import torch

from resnet import BasicBlock, ResNet18

# CIFAR-10 image shape is (batch_size, 3, 32, 32)
block = BasicBlock(3, 64, stride=2)
x = torch.randn(1, 3, 32, 32)  # CIFAR-10 input: 1 image, 3 channels, 32x32
y = block(x)

dot = make_dot(y, params=dict(block.named_parameters()))
dot.render("./compute-graph/basicblock_cifar10", format="png")  # writes basicblock_cifar10.png
# or: print(dot) / dot.view()
```
